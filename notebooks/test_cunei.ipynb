```python
from IPython.display import HTML, display

def print_table(text):
    """
    Imprime texto asegurando fuente monoespaciada y soporte cuneiforme
    en entornos de Notebook, evitando el desalineamiento.
    """
    html_output = f"""
    <div style="
        font-family: 'Noto Sans Cuneiform', 'Courier New', monospace !important;
        white-space: pre;
        line-height: 1.2em;
        font-size: 1.1em;
        background-color: #f8f9fa;
        padding: 10px;
        border-left: 3px solid #ccc;
        display: block;
        color: #1a1a1a;
    ">{text}</div>
    """
    display(HTML(html_output))
```

```python
from mesomath import Bcap as bc
tabla_texto = "\n".join([_[0] for _ in bc.metro_generator(mmin="1 sila3", mmax="10 sila3", step="1 sila3", cuneiform=True)])
```

```python
print_table( tabla_texto )
```
